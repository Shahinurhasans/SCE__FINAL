```python
import os, json
import numpy as np
import pyarrow as pa
import pyarrow.parquet as pq
import pandas as pd
```

```python
TRAIN_IN = "/kaggle/input/datasets/zerotoinfinitya/raw-dataset/train_dataset.parquet"
VAL_IN = "/kaggle/input/datasets/zerotoinfinitya/raw-dataset/val_dataset.parquet"
TEST_IN = "/kaggle/input/datasets/zerotoinfinitya/raw-dataset/test_dataset.parquet"

TRAIN_OUT = "/kaggle/working/train_with_Label.parquet"
VAL_OUT = "/kaggle/working/val_with_Label.parquet"
TEST_OUT = "/kaggle/working/test_with_Label.parquet"

LABELS = ["access-control", "arithmetic", "other",
          "reentrancy", "safe", "unchecked-calls"]
```

```python
def add_labels(df):
    names = []
    for numbers in df["slither"]:
        numbers = [] if numbers is None else list(numbers)
        names.append([LABELS[n] for n in numbers])
    df["labels"] = [", ".join(x) for x in names]
    return df
```

```python
train = pd.read_parquet(TRAIN_IN)
train = add_labels(train)
train.to_parquet(TRAIN_OUT, index=False)
```

```python
val = pd.read_parquet(VAL_IN)
val = add_labels(val)
val.to_parquet(VAL_OUT, index=False)
```

```python
test = pd.read_parquet(TEST_IN)
test = add_labels(test)
test.to_parquet(TEST_OUT, index=False)
```
